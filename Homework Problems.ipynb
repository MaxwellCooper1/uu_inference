```python
import pandas as pd
import numpy as np

data=pd.read_csv('./data/ames_prices.csv')
data.head()
```

```python
pd.set_option('display.max_rows', 100)
data.head().T
```

```python
data['Central.Air']
```

```python
data['Central.Air']=np.where(data['Central.Air']=="Y",1,0)
```

```python
data['Central.Air']
```

```python
# Estimate propotion of central air=yes

sample_estimate=sum(data['Central.Air'])*(1/len(data))
sample_estimate
```

## Part B

```python
rng=np.random.default_rng(seed=51)
k=5000
ninv=1/len(data)

bootstrap_propotion=[]
for i in range(k):
    air=rng.choice(data['Central.Air'], size=len(data), replace=True)
    prop=sum(air)*ninv
    bootstrap_propotion.append(prop)

average_prop=round(np.mean(bootstrap_propotion),3)

print(f'After {k} bootstraped samples, the average proportion of central air is about {average_prop} with rng seed 51')
```

## Part C

```python
bootstrap_standard_error=np.std(bootstrap_propotion, ddof=1)
bootstrap_standard_error
```

## Part D

```python
print(f'90% confidence interval [{sample_estimate-1.645*bootstrap_standard_error},{sample_estimate+1.645*bootstrap_standard_error}]')

print(f'95% confidence interval [{sample_estimate-1.96*bootstrap_standard_error},{sample_estimate+1.96*bootstrap_standard_error}]')

print(f'99% confidence interval [{sample_estimate-2.576*bootstrap_standard_error},{sample_estimate+2.576*bootstrap_standard_error}]')
```

## Part E

$H_0:p=.9 \qquad \alpha=0.1,\quad 0.05,\quad 0.01$

At every level for $\alpha$, we reject the null hypothesis. In each  $(1-\alpha)$ confidence interval, .9 is outside of the interval, so there is enough statistically significant evidence to reject the null. Here, we can claim that the probability of having central air is larger than 90%

```python
transormed_area=np.arcsinh(data['area'])
transformed_price=np.arcsinh(data['price'])
```

```python
sample_corr=np.corrcoef(transormed_area, transformed_price)
sample_corr[0][1]
```

## Part a

```python
rng=np.random.default_rng(seed=51)
k=5000
ninv=1/len(data)


bootstrap_corr=[]
for i in range(k):
    picked_index=rng.choice(range(len(data)), size=len(data), replace=True)
    bootstraped_price=transformed_price[picked_index]
    bootstrapped_area=transormed_area[picked_index]
    sample_bootstrap_corr=np.corrcoef(bootstraped_price,bootstrapped_area)
    bootstrap_corr.append(sample_bootstrap_corr[0][1])
    
corr_mean=np.mean(bootstrap_corr)

print(f'After {k} bootstraped samples, the average corr of transformed price and area is about {corr_mean} with rng seed 51')
```

## Part B

```python
bootstrap_standard_error=np.std(bootstrap_corr, ddof=1)
bootstrap_standard_error
```

## Part C

```python
print(f'90% confidence interval [{sample_corr[0][1]-1.645*bootstrap_standard_error},{sample_corr[0][1]+1.645*bootstrap_standard_error}]')

print(f'95% confidence interval [{sample_corr[0][1]-1.96*bootstrap_standard_error},{sample_corr[0][1]+1.96*bootstrap_standard_error}]')

print(f'99% confidence interval [{sample_corr[0][1]-2.576*bootstrap_standard_error},{sample_corr[0][1]+2.576*bootstrap_standard_error}]')
```

## Part D

$H_0:\rho=0 \quad H_1:\rho\neq 0 \qquad \alpha=0.1,\quad 0.05,\quad 0.01$

At every level for $\alpha$, we reject the null hypothesis. In each  $(1-\alpha)$ confidence interval, 0 is outside of the interval, so there is enough statistically significant evidence to reject the null. Here, we can claim that there is a correlation between the area of a house and its price even though we had to transform our variables first.

## Part E

We had to keep the two variables paired because it would not make sense for us to calculate the correlation of two variables from two random different houses. The analysis only works because we are analyzing both the area and sales price of the same house.

```python
death=pd.read_csv('./data/metabric.csv')
death.head()
```

```python
sample_survival=np.mean(death['Overall Survival (Months)'])
sample_survival
```

## Part a

```python
rng=np.random.default_rng(seed=51)
k=5000
ninv=1/len(death)


bootstrap_survival=[]
for i in range(k):
    survival=rng.choice(death['Overall Survival (Months)'], size=len(death), replace=True)
    mean_survival=np.mean(survival)
    bootstrap_survival.append(mean_survival)
    
survival_average=round(np.mean(bootstrap_survival),3)


print(f'After {k} bootstraped samples, the average number of months for survival is about {survival_average} with rng seed 51')
```

## Part B

```python
bootstrap_standard_error=np.std(bootstrap_survival, ddof=1)
bootstrap_standard_error
```

## Part C

```python
print(f'90% confidence interval [{sample_survival-1.645*bootstrap_standard_error},{sample_survival+1.645*bootstrap_standard_error}]')

print(f'95% confidence interval [{sample_survival-1.96*bootstrap_standard_error},{sample_survival+1.96*bootstrap_standard_error}]')

print(f'99% confidence interval [{sample_survival-2.576*bootstrap_standard_error},{sample_survival+2.576*bootstrap_standard_error}]')
```

## Part D

$H_0:\mu=120 \quad H_1:\mu\neq 120 \qquad \alpha=0.1,\quad 0.05,\quad 0.01$

At every level for $\alpha$, we reject the null hypothesis. In each  $(1-\alpha)$ confidence interval, 120 is outside of the interval, so there is enough statistically significant evidence to reject the null. Here, we can claim the true value for average survival time is more than 120 months.

## Part E

We are 95% confident that the true value for survival months as time of death or last doctor's visit is captured in the range [124.8, 133.3] months. or [10.4, 11.1] years. Although, this is not an accurate measure of lifespan after treatment since one could have a final doctor's appointment months or even years before their eventual death and those months would not be accounted for with this data. We do know that this is at least not an overestimate since all patients were alive at least up to the recorded time.

```python
sleep=pd.read_csv('./data/nhanes_data_17_18.csv')
pd.set_option('display.max_rows', 200)
sleep.head().T
```

```python
sleep.shape
```

```python
filtersleep=sleep[(sleep['AgeInYearsAtScreening']>=18) & (sleep['SleepHoursWeekdaysOrWorkdays'].notna())]
sleep=filtersleep
```

```python
sleep.shape
```

```python
sample_sleep=np.mean(sleep['SleepHoursWeekdaysOrWorkdays'])
sample_sleep
```

## Part A

```python
rng=np.random.default_rng(seed=51)
k=5000
ninv=1/len(sleep)

bootstrap_sleep=[]
for i in range(k):
    sleep_bootstrap=rng.choice(sleep['SleepHoursWeekdaysOrWorkdays'], size=len(sleep), replace=True)
    ave=sum(sleep_bootstrap)*ninv
    bootstrap_sleep.append(ave)

average_sleep=round(np.mean(bootstrap_sleep),3)

print(f'After {k} bootstraped samples, the average number of hours of sleep is about {average_sleep} with rng seed 51')
```

## Part B

```python
bootstrap_standard_error=np.std(bootstrap_sleep, ddof=1)
bootstrap_standard_error
```

## Part C

```python
print(f'90% confidence interval [{sample_sleep-1.645*bootstrap_standard_error},{sample_sleep+1.645*bootstrap_standard_error}]')

print(f'95% confidence interval [{sample_sleep-1.96*bootstrap_standard_error},{sample_sleep+1.96*bootstrap_standard_error}]')

print(f'99% confidence interval [{sample_sleep-2.576*bootstrap_standard_error},{sample_sleep+2.576*bootstrap_standard_error}]')
```

## Part D

$H_0:\mu=8 \quad H_1:\mu\neq 8 \qquad \alpha=0.1,\quad 0.05,\quad 0.01$

At every level for $\alpha$, we reject the null hypothesis. In each  $(1-\alpha)$ confidence interval, 8 is outside of the interval, so there is enough statistically significant evidence to reject the null. Here, we can claim the true value for average number of hours of sleep on the weekend or workday is less than 8 hours a night.

## Part E

While there is a statistical significance, it is subjective whether the results pose practical significance. Even at 99% confidence, the lower bound for average sleep is above 7.5 so even the worst case has the true value less than half an hour off the prediction. Now, if you're me in the morning, that extra half an hour makes a world of difference, but from a planing perspective, that extra half an hour in the morning might not contribute much.
